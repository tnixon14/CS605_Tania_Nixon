When we divide with 0 we have two cases:
>Case 1: 7/0<br>
>Case 2: 0/0

For Case 1, any integer divided by 0 will result in an error as it is impossible to compute.<br>
For Case 2, 0 divided by 0 also results in an error since it would be considered undefined.

For multiplication of numbers there are three cases of note:
>Case 3: 999999999.0 * 999999999.0 = 9.99999998e+17<br>
Case 4: 0.0000009 * 0.0000009 = 8.099999999999999e-13<br>
Case 6: 8.0 * -6.0 = -48.0<br>

The calculator outputs very large and small numbers in exponential form and is even able to work with negative integers.

Adding and dividing by negative integers
>Case 7: -9.0 + -11.0 = -20.0<br>
>Case 8: -12/345678 = -3.471429699643224e-06


All cases along with standard positive cases and observations outputted expected results. All simple calculations will work except for two cases:<br>
1) When both numbers are 0<br>
2) When the second number is 0 and the operation is division
